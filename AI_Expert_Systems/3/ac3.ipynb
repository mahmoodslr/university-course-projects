```python
from collections import deque
import copy
```

```python
class AC3:
    def __init__(self, variables, domains, constraints):
        self.variables = variables
        self.domains = {var: list(values) for var, values in domains.items()}
        self.constraints = constraints
        self.queue = deque()
        self.arcs = []
        
        # ساختن لیست همه قوس‌ها (برای گراف کامل، همه زوج‌ها دوطرفه)
        for i in range(len(variables)):
            for j in range(len(variables)):
                if i != j:
                    self.arcs.append((variables[i], variables[j]))
        
        # پر کردن صف با همه قوس‌ها
        for (X, Y) in self.arcs:
            self.queue.append((X, Y))
```

```python
def revise(self, X, Y):
    revised = False
    if (X, Y) in self.constraints:
        constraint = self.constraints[(X, Y)]
    elif (Y, X) in self.constraints:
        constraint = lambda x, y: self.constraints[(Y, X)](y, x)
    else:
        return False
    
    new_domain = []
    for x in self.domains[X]:
        has_support = False
        for y in self.domains[Y]:
            if constraint(x, y):
                has_support = True
                break
        if has_support:
            new_domain.append(x)
        else:
            revised = True
    
    if revised:
        self.domains[X] = new_domain
    return revised

AC3.revise = revise
```

```python
def run(self):
    iteration = 0
    print("شروع الگوریتم AC-3...")
    print(f"دامنه‌های اولیه: {self.domains}")
    print(f"تعداد قوس‌های اولیه: {len(self.queue)}")
    print("-" * 50)
    
    while self.queue:
        X, Y = self.queue.popleft()
        iteration += 1
        print(f"تکرار {iteration}: بررسی قوس ({X} -> {Y})")
        print(f"  دامنه {X} قبل: {self.domains[X]}")
        
        if self.revise(X, Y):
            print(f"  ✂️  دامنه {X} تغییر کرد! دامنه جدید: {self.domains[X]}")
            
            if not self.domains[X]:
                print(f"  ❌ دامنه {X} خالی شد! ناسازگاری تشخیص داده شد.")
                return False
            
            for Z in self.variables:
                if Z != Y and Z != X:
                    self.queue.append((Z, X))
                    print(f"  ➕ قوس ({Z} -> {X}) به صف اضافه شد")
        else:
            print(f"  ✅ دامنه {X} بدون تغییر")
        
        print(f"  دامنه‌های فعلی: {self.domains}")
        print("-" * 30)
    
    print("✅ الگوریتم پایان یافت. همه قوس‌ها سازگارند.")
    return True

AC3.run = run
```

```python
def get_solution(self):
    if all(len(self.domains[var]) > 0 for var in self.variables):
        return {var: self.domains[var][0] for var in self.variables}
    return None

AC3.get_solution = get_solution
```

```python
# متغیرها (رأس‌های نقشه)
variables = ['X1', 'X2', 'X3']

# دامنه رنگ‌ها (مشابه قبل)
domains = {
    'X1': ['R'],
    'X2': ['R'],
    'X3': ['R', 'G']
}

# تابع محدودیت: رنگ‌ها باید متفاوت باشند
def constraint_neq(x, y):
    return x != y

# محدودیت‌ها برای گراف کامل (همه جفت‌ها)
constraints = {
    ('X1', 'X2'): constraint_neq,
    ('X2', 'X1'): constraint_neq,
    ('X2', 'X3'): constraint_neq,
    ('X3', 'X2'): constraint_neq,
    ('X1', 'X3'): constraint_neq,  # ✅ اینجا تغییر کرد (قبلاً constraint_eq بود)
    ('X3', 'X1'): constraint_neq,  # ✅ اینجا تغییر کرد (قبلاً constraint_eq بود)
}

print("=" * 60)
print("مسئله رنگ‌آمیزی نقشه با گراف کامل (مثلث)")
print("=" * 60)
print(f"متغیرها: {variables}")
print(f"دامنه‌ها: {domains}")
print("محدودیت‌ها (همه جفت‌ها باید متفاوت باشند):")
print("  X1 != X2")
print("  X2 != X3")
print("  X1 != X3")
print("=" * 60)
```

```python
ac3 = AC3(variables, domains, constraints)
has_solution = ac3.run()

print("\n" + "=" * 60)
print("نتیجه نهایی:")
print("=" * 60)

if has_solution:
    solution = ac3.get_solution()
    print(f"✅ مسئله رنگ‌آمیزی جواب دارد: {solution}")
else:
    print("❌ مسئله رنگ‌آمیزی جواب ندارد!")
    print("اثبات: دامنه یکی از رأس‌ها (X1) خالی شد.")
    print(f"دامنه‌های نهایی: {ac3.domains}")
```

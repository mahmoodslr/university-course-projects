```python
import copy

# حالت هدف
GOAL = [[1, 2, 3],
        [4, 5, 6],
        [7, 8, 0]]
```

```python
def find_empty(state):
    """پیدا کردن موقعیت خانه خالی (0)"""
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
    return None

def manhattan_distance(state):
    """محاسبه فاصله منهتن از حالت هدف"""
    distance = 0
    for i in range(3):
        for j in range(3):
            val = state[i][j]
            if val != 0:
                target_i, target_j = (val-1) // 3, (val-1) % 3
                distance += abs(i - target_i) + abs(j - target_j)
    return distance
```

```python
def get_neighbors(state):
    """تولید همه همسایه‌های ممکن (۴ حرکت) به ترتیب: بالا، پایین، چپ، راست"""
    i, j = find_empty(state)
    neighbors = []
    moves = [(-1,0), (1,0), (0,-1), (0,1)]
    for di, dj in moves:
        ni, nj = i+di, j+dj
        if 0 <= ni < 3 and 0 <= nj < 3:
            new_state = copy.deepcopy(state)
            new_state[i][j], new_state[ni][nj] = new_state[ni][nj], new_state[i][j]
            neighbors.append(new_state)
    return neighbors
```

```python
def print_step(iteration, state, improvement):
    """چاپ وضعیت پذیرفته شده در هر تکرار"""
    print(f"--- [ تکرار: {iteration} ] ---")
    print(f"بهبود: {improvement}")
    for row in state:
        print("  " + " ".join(str(cell) if cell != 0 else "_" for cell in row))
    print(f"فاصله منهتن فعلی: {manhattan_distance(state)}")
    print("-" * 40)
```

```python
def simple_hill_climbing(initial_state, verbose=True):
    """
    الگوریتم تپه‌نوردی ساده (Simple Hill Climbing)
    
    پارامترها:
        initial_state: حالت اولیه (لیست ۳×۳)
        verbose: اگر True باشد، جزئیات每一步 را چاپ می‌کند.
    
    خروجی:
        (state, cost, steps)  →  حالت نهایی، هزینه نهایی، تعداد گام‌ها
    """
    current_state = copy.deepcopy(initial_state)
    current_cost = manhattan_distance(current_state)
    
    if verbose:
        print("=" * 60)
        print("الگوریتم تپه‌نوردی ساده (Simple Hill Climbing)")
        print("=" * 60)
        print("\nحالت اولیه:")
        for row in initial_state:
            print("  " + " ".join(str(cell) if cell != 0 else "_" for cell in row))
        print(f"فاصله منهتن اولیه: {current_cost}")
        print("=" * 60)
    
    iteration = 0
    
    while True:
        neighbors = get_neighbors(current_state)
        
        for neighbor in neighbors:
            neighbor_cost = manhattan_distance(neighbor)
            
            if neighbor_cost < current_cost:
                improvement = current_cost - neighbor_cost
                
                current_state = neighbor
                current_cost = neighbor_cost
                iteration += 1
                
                if verbose:
                    print_step(iteration, current_state, f"{improvement:+d}")
                
                break
        else:
            if verbose:
                if current_cost == 0:
                    print(f"🎯 جواب کامل پیدا شد! (هزینه: {current_cost})")
                else:
                    print(f"⛰️ بهینه محلی! هیچ همسایه‌ای بهتر از حالت فعلی وجود ندارد.")
                    print(f"فاصله منهتن نهایی: {current_cost}")
            return current_state, current_cost, iteration
```

```python
def print_puzzle(state):
    """چاپ زیبای پازل"""
    for row in state:
        print(" ".join(str(cell) if cell != 0 else " " for cell in row))
    print("فاصله منهتن:", manhattan_distance(state))
```

```python
# حالت پیچیده (سخت)
initial = [[8, 6, 7],
           [2, 5, 4],
           [3, 0, 1]]

# اجرا
final_state, final_cost, steps = simple_hill_climbing(initial, verbose=True)

# نتیجه
print("\n" + "=" * 40)
print("نتیجه نهایی:")
print_puzzle(final_state)
print(f"گام‌ها: {steps}")
print(f"هزینه: {final_cost}")
print("✅ موفق" if final_cost == 0 else "❌ شکست")
```
